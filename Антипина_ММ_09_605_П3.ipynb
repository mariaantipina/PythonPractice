```python
class Member:
    def __init__(self, name, subscription_type, months):
        self.name = name
        self.subscription_type = subscription_type
        self.months = months

    def __str__(self):
        # Человекочитаемое представление — для print(), f-строк, отчётов
        return (f'Клиент "{self.name}" '
                f'(абонемент "{self.subscription_type}", {self.months} мес.)')

    def __repr__(self):
        # Техническое представление — для отладчика, списков, eval()
        return (f"Member({self.name!r}, "
                f"{self.subscription_type!r}, {self.months!r})")


# Проверка
member = Member("Иван Петров", "Премиум", 12)

print(member)             # Клиент "Иван Петров" (абонемент "Премиум", 12 мес.)
print(repr(member))       # Member('Иван Петров', 'Премиум', 12)

# В списке Python использует __repr__ для элементов
members = [member, Member("Мария Сидорова", "Стандарт", 6)]
print(members)
# [Member('Иван Петров', 'Премиум', 12), Member('Мария Сидорова', 'Стандарт', 6)]

# А str() явно — использует __str__
print(str(member))        # Клиент "Иван Петров" (абонемент "Премиум", 12 мес.)

"""
РАЗНИЦА МЕЖДУ __str__ И __repr__:

ПРАВИЛА ХОРОШЕГО ТОНА:
1. __repr__ должен быть у ЛЮБОГО класса — это «лицо» объекта для отладки.
2. __str__ нужен, только если хочется более «человечное» представление,
   чем даёт __repr__. Если __str__ не задан, Python использует __repr__.
3. В контейнерах (список, кортеж, словарь) Python ВСЕГДА использует __repr__,
   даже если у элементов есть __str__:

       print([member])   # [Member('Иван Петров', 'Премиум', 12)]
       print(member)     # Клиент "Иван Петров" (абонемент "Премиум", 12 мес.)

4. При написании __repr__ принято использовать !r (repr) для строковых
   полей — тогда кавычки и экранирование подставятся автоматически.

ИТОГ: __str__ — «витрина» для человека, __repr__ — «паспорт» для разработчика.
"""

print()
```

```python
class Member:
    def __init__(self, name, subscription_type):
        self.name = name
        self.subscription_type = subscription_type

    def __eq__(self, other):
        # Два клиента равны, если совпадают имя и тип абонемента
        if not isinstance(other, Member):
            return NotImplemented
        return (self.name == other.name
                and self.subscription_type == other.subscription_type)

    def __hash__(self):
        # Хэш должен считаться по ТЕМ ЖЕ полям, что и __eq__,
        # иначе нарушится контракт "равные объекты → равные хэши"
        return hash((self.name, self.subscription_type))

    def __repr__(self):
        return f"Member({self.name!r}, {self.subscription_type!r})"

a = Member("Иван Петров", "Премиум")
b = Member("Иван Петров", "Премиум")   # тот же клиент по данным
c = Member("Мария Сидорова", "Стандарт")

print(a == b)              # True  — совпадают имя и абонемент
print(a == c)              # False — разные клиенты

# Множество: дубликат НЕ добавляется
members = {a, b, c}
print(members)
# {Member('Иван Петров', 'Премиум'), Member('Мария Сидорова', 'Стандарт')}
print(len(members))        # 2, а не 3

# Словарь: a и b — один и тот же ключ
visits = {a: 10}
visits[b] = 20             # перезапишет значение по тому же ключу
print(visits)              # {Member('Иван Петров', 'Премиум'): 20}

"""
ПОЧЕМУ ПРИ ПЕРЕОПРЕДЕЛЕНИИ __eq__ НУЖНО ПЕРЕОПРЕДЕЛИТЬ И __hash__:

В Python действует ЖЕЛЕЗНЫЙ КОНТРАКТ для хэшируемых объектов:

    Если a == b, то ОБЯЗАТЕЛЬНО hash(a) == hash(b).

Нарушение этого правила ломает работу set и dict, потому что они устроены так:
1. По хэшу вычисляется «корзина», в которой искать объект.
2. Внутри корзины уже идёт сравнение через ==.

ЧТО ПРОИЗОЙДЁТ, ЕСЛИ ПЕРЕОПРЕДЕЛИТЬ ТОЛЬКО __eq__, НО НЕ __hash__:

Сам Python автоматически обнуляет __hash__ у класса, в котором вы
определили __eq__. Формально это выглядит так:

    class Member:
        __hash__ = None   # ← Python подставляет это неявно

В результате объекты класса становятся НЕХЭШИРУЕМЫМИ:

    a = Member("Иван", "Премиум")
    {a}                     # TypeError: unhashable type: 'Member'
    {a: 1}                  # TypeError: unhashable type: 'Member'

Python делает это специально, чтобы защитить вас от логических ошибок:
если бы он оставил старый __hash__ (по id объекта), то два РАВНЫХ клиента
получили бы РАЗНЫЕ хэши — и в set/dict они попали бы как два разных элемента.
Именно та проблема, которую мы решаем в этой задаче, осталась бы.

ПРАВИЛА:
1. Переопределили __eq__ → почти всегда переопределяйте и __hash__.
2. Хэш считайте по ТЕМ ЖЕ (или по подмножеству тех же) полей, что и __eq__.
   Обычно: hash((self.field1, self.field2, ...)).
3. Если объект по смыслу изменяемый (mutable) и не должен быть ключом —
   можно оставить __hash__ = None осознанно (как у list и dict).
4. Если объект неизменяемый (immutable) и логически «значимый» (как наш
   клиент) — смело делайте его хэшируемым: он станет удобным ключом
   для словарей и элементом множеств.

ИТОГ: __eq__ без __hash__ = объект, который нельзя положить в set или dict.
      Переопределяйте их ПАРОЙ и считайте хэш по тем же полям, что и равенство.
"""

print()
```

```python
class Group:
    def __init__(self, members=None):
        # members=None (а не []) — классический приём, чтобы не делить
        # один и тот же список между всеми объектами по умолчанию
        self.members = list(members) if members else []

    def add_member(self, member):
        self.members.append(member)

    def __add__(self, other):
        # Объединение двух групп → НОВАЯ группа
        if not isinstance(other, Group):
            return NotImplemented
        return Group(self.members + other.members)

    def __len__(self):
        # len(g) → количество клиентов в группе
        return len(self.members)

    def __repr__(self):
        return f"Group({self.members!r})"

g1 = Group(["Иван", "Мария", "Пётр"])
g2 = Group(["Алексей", "Ольга"])

print(len(g1))          # 3
print(len(g2))          # 2

g3 = g1 + g2            # объединение
print(len(g3))          # 5
print(g3)               # Group(['Иван', 'Мария', 'Пётр', 'Алексей', 'Ольга'])

# Убедимся, что исходные группы НЕ изменились
print(len(g1))          # 3
print(len(g2))          # 2
print(g1 is g3)         # False — это разные объекты

"""
ПОЧЕМУ __add__ ВОЗВРАЩАЕТ НОВЫЙ ОБЪЕКТ, А НЕ ИЗМЕНЯЕТ СУЩЕСТВУЮЩИЙ:

1. Контракт оператора +.
   В Python (и математике) a + b означает «взять a и b и получить
   НЕЧТО ТРЕТЬЕ». Никто не ожидает, что после c = a + b изменится сам a.

       x = 2
       y = 3
       z = x + y     # x и y не изменились, z = 5

   Если бы Group.__add__ менял self, поведение ломало бы интуицию:

       g3 = g1 + g2  # ожидаем НОВУЮ группу,
                     # а получили бы "g1 теперь содержит и клиентов g2"

2. Ссылочная прозрачность.
   Если g1 + g2 мутирует g1, то где-то в другом месте программы,
   где на g1 ещё кто-то ссылается, данные внезапно «поедут».
   Это классический источник трудноуловимых багов.

3. Аналогия со встроенными типами.
       [1, 2] + [3, 4]   # → [1, 2, 3, 4], исходные списки целы
       "ab" + "cd"       # → "abcd", исходные строки целы (они вообще неизменяемы)
       (1, 2) + (3,)     # → (1, 2, 3), исходные кортежи целы
   Наш Group.__add__ ведёт себя ТОЧНО ТАК ЖЕ — как встроенные типы.
   Это и есть смысл магических методов: «притвориться» встроенным типом.

4. А если хочется изменить на месте?
   Для этого есть ДРУГОЙ оператор — +=, и магический метод __iadd__:

       def __iadd__(self, other):
           self.members += other.members
           return self

   Тогда g1 += g2 изменит g1. Но + и += — это разные операторы,
   и вести себя они должны по-разному.

"""

print()
```

```python
class Member:
    def __init__(self, name, months):
        self.name = name
        # ВАЖНО: сохраняем в "закрытое" поле _months, а не в months,
        # иначе получим бесконечную рекурсию (setter вызывает сам себя)
        self.months = months   # ← идёт через setter, проверка сработает уже тут

    @property
    def months(self):
        # Геттер: срабатывает при чтении member.months (без скобок!)
        return self._months

    @months.setter
    def months(self, value):
        # Сеттер: срабатывает при записи member.months = ...
        # 1) Проверка типа: bool — тоже int в Python, поэтому исключаем отдельно
        if not isinstance(value, int) or isinstance(value, bool):
            raise ValueError(
                f"Срок абонемента должен быть целым числом, "
                f"получено {type(value).__name__}: {value!r}"
            )
        # 2) Проверка значения
        if value <= 0:
            raise ValueError(
                f"Срок абонемента должен быть строго больше нуля, "
                f"получено: {value}"
            )
        # 3) Всё в порядке — сохраняем
        self._months = value

    def __repr__(self):
        return f"Member({self.name!r}, {self.months!r})"


# Проверка
m = Member("Иван Петров", 12)
print(m.months)            # 12 — читается как обычный атрибут, без скобок

m.months = 6
print(m.months)            # 6

"""
ЧЕМ @property УДОБНЕЕ КЛАССИЧЕСКИХ ГЕТТЕРОВ И СЕТТЕРОВ:

Классический стиль:

    class Member:
        def get_months(self):
            return self._months
        def set_months(self, value):
            if value <= 0:
                raise ValueError(...)
            self._months = value

    m.set_months(12)
    print(m.get_months())

Проблемы этого стиля:
1. Вызывающему коду приходится писать методы (со скобками) вместо
   простого обращения к атрибуту.
2. Если потом вы захотите добавить проверку к уже существующему полю,
   придётся менять ВЕСЬ код, который делал m.months = ...
   на m.set_months(...) — это ломающее изменение API.

Преимущества @property:
1. СИНТАКСИС АТРИБУТА. Снаружи всё выглядит как обычное поле:

       m.months          # читаем
       m.months = 12     # пишем
       m.months += 1     # тоже работает (через getter + setter)

   Никаких get_/set_ — код чище и естественнее.

3. ЛОГИКА ИНКАПСУЛИРОВАНА. Проверки, преобразования, ленивое вычисление —
   всё спрятано внутри класса, наружу торчит только чистый интерфейс.

4. МОЖНО СДЕЛАТЬ ТОЛЬКО ДЛЯ ЧТЕНИЯ. Если определить @property без сеттера,
   поле становится read-only:

       @property
       def id(self):
           return self._id

       m.id = 5   # AttributeError: can't set attribute

ЧТО ТАКОЕ «ВЫЧИСЛЯЕМЫЙ АТРИБУТ» (computed attribute):

Это атрибут, который НЕ хранится в объекте, а ВЫЧИСЛЯЕТСЯ на лету
при каждом обращении. Снаружи он выглядит как обычное поле, но внутри
за ним стоит код.

Пример — «статус абонемента»:

    class Member:
        def __init__(self, name, months):
            self.name = name
            self.months = months

        @property
        def is_active(self):
            # Никакого self._is_active нет — значение вычисляется
            return self.months > 0

        @property
        def months_left_until_expiry(self):
            return max(0, self.months - 12)

    m = Member("Иван", 14)
    print(m.is_active)                 # True  — вычислено только что
    print(m.months_left_until_expiry)  # 2

Ключевые свойства вычисляемого атрибута:
- хранится ТОЛЬКО логика, а не значение;
- значение всегда АКТУАЛЬНО (не нужно синхронизировать с другими полями);
- экономит память (не держим дублирующие данные);
- читается как обычное поле — m.is_active, без скобок.

Минус: вычисление происходит КАЖДЫЙ раз при чтении. Если оно тяжёлое —
можно закешировать через @functools.cached_property (Python 3.8+).

ИТОГ:
- @property даёт «магию»: снаружи — атрибут, внутри — метод с логикой.
- Код-потребитель остаётся простым: m.months, m.months = 12.
- Проверки и вычисления прячутся внутри класса — это и есть инкапсуляция.
- Вычисляемый атрибут — это @property, который не хранит значение, а
  каждый раз его считает из других данных. Снаружи — неотличим от поля.
"""

print()
```

```python
class Member:
    def __init__(self, name, subscription_type, months):
        self.name = name
        self.subscription_type = subscription_type
        self.months = months

    def show(self):
        # ОБЫЧНЫЙ метод: работает с КОНКРЕТНЫМ объектом через self
        print(f"Клиент: {self.name} ({self.subscription_type}, {self.months} мес.)")

    @classmethod
    def from_string(cls, data):
        # КЛАССОВЫЙ метод: получает КЛАСС через cls, а не объект
        # Работает как альтернативный конструктор
        name, subscription_type, months = data.split(";")
        return cls(name, subscription_type, int(months))

    @staticmethod
    def is_valid_data(data):
        # СТАТИЧЕСКИЙ метод: ни self, ни cls — просто утилита
        parts = data.split(";")
        if len(parts) != 3:
            return False
        try:
            int(parts[2])
        except ValueError:
            return False
        return True

    def __repr__(self):
        return f"Member({self.name!r}, {self.subscription_type!r}, {self.months!r})"

# 1. Обычное создание
m1 = Member("Мария Сидорова", "Стандарт", 6)
m1.show()

# 2. Создание через классовый метод — как альтернативный конструктор
raw = "Иван Петров;Премиум;12"
m2 = Member.from_string(raw)
m2.show()
print(repr(m2))

# 3. Статический метод — утилита, не требует ни объекта, ни класса
print(Member.is_valid_data("Иван;Премиум;12"))    # True
print(Member.is_valid_data("Иван"))              # False

# Статический метод можно вызывать и от экземпляра — но это тот же метод
print(m2.is_valid_data("А;Б;3"))                  # True

"""
РАЗНИЦА МЕЖДУ ТРЕМЯ ВИДАМИ МЕТОДОВ:

┌────────────────┬───────────────┬──────────────────────────────────────────┐
│ Вид            │ 1-й аргумент  │ Назначение                               │
├────────────────┼───────────────┼──────────────────────────────────────────┤
│ обычный метод  │ self          │ Работа с КОНКРЕТНЫМ объектом
         │
│ @classmethod   │ cls (класс)   │ Альтернативный конструктор, фабрика,     │
│                │               │ работа с классом в целом
           │
│ @staticmethod  │ —             │ Утилита, не зависящая ни от объекта,     │
│                │               │ ни от класса                             │
└────────────────┴───────────────┴──────────────────────────────────────────┘
КОГДА ЧТО ИСПОЛЬЗОВАТЬ:

Обычный метод (self):
- Действие над конкретным объектом: m.show(), m.renew(), g.add_member(x).
- Читает/меняет поля экземпляра.
- Признак: в теле нужен self.<что-то>.

@classmethod (cls):
- Альтернативный конструктор: from_string, from_dict, from_json, from_db_row.
- Фабричный метод, который может вернуть объект класса или его подкласса.
- Действие над всем классом: подсчёт всех созданных объектов, настройка класса.
- Признак: в теле нужен cls(...) или cls.<атрибут класса>.

@staticmethod:
- Чистая утилита, не трогающая ни self, ни cls.
- Валидация, парсинг, преобразование, математика.
- Логически «принадлежит» классу, но технически — обычная функция.
- Признак: тело не использует ни self, ни cls; метод мог бы быть
  свободной функцией, но хочется держать его рядом с классом.
"""

print()
```

```python
class Member:
    # КЛАССОВЫЙ атрибут — один на весь класс, общий для всех объектов
    count = 0

    def __init__(self, name):
        self.name = name                     # ЭКЗЕМПЛЯРНЫЙ атрибут
        Member.count += 1                    # увеличиваем ОБЩИЙ счётчик

    def __repr__(self):
        return f"Member({self.name!r})"

a = Member("Иван")
b = Member("Мария")
c = Member("Пётр")

print(Member.count)      # 3 — общий счётчик класса
print(a.count)           # 3 — виден через экземпляр (ищется в классе)

# Демонстрация: изменение через экземпляр НЕ трогает классовый атрибут
a.count = 100

print(a.count)           # 100 — появился ЛИЧНЫЙ атрибут экземпляра a
print(b.count)           # 3   — у b по-прежнему классовый
print(Member.count)      # 3   — классовый не изменился!

# Куда реально записалось значение:
print(a.__dict__)        # {'name': 'Иван', 'count': 100}  ← личный атрибут
print(Member.__dict__['count'])   # 3  ← классовый атрибут на месте

# А если изменить через КЛАСС — увидят все, кто не имеет своего count
Member.count = 999
print(b.count)           # 999
print(c.count)           # 999
print(a.count)           # 100 — у a свой личный, он "перекрывает" классовый
print(Member.count)      # 999

"""
РАЗНИЦА МЕЖДУ КЛАССОВЫМИ И ЭКЗЕМПЛЯРНЫМИ АТРИБУТАМИ:

┌───────────────────┬──────────────────────────┬──────────────────────────┐
│                   │ Классовый атрибут        │ Экземплярный атрибут     │
├───────────────────┼──────────────────────────┼──────────────────────────┤
│ Где объявлен      │ в теле класса            │ в __init__ (или где-то   │
│                   │ (вне методов)            │ ещё через self.x = ...)
  │
│ Хранится в        │ Member.__dict__          │ m.__dict__
     │
│ Общий для         │ ВСЕХ объектов класса     │ только ЭТОГО объекта
│
│ Изменение через   │ меняет у всех            │ создаёт ЛИЧНУЮ копию,    │
│ экземпляр         │                          │ классовый не трогает
  │
│ Типичное примен.  │ счётчики, настройки,     │ имя, возраст, состояние  │
│                   │ общие константы          │ конкретного объекта      │
└───────────────────┴──────────────────────────┴──────────────────────────┘
ИТОГ:
- Классовый атрибут живёт в классе, общий для всех; обращаться и менять
  его лучше через имя класса: Member.count += 1.
- Экземплярный атрибут живёт в объекте, у каждого свой.
- При ЧТЕНИИ Python идёт: экземпляр → класс → родители → метакласс.
- При ЗАПИСИ через obj.attr = x всегда создаётся атрибут ЭКЗЕМПЛЯРА,
  классовый остаётся нетронутым (кроме случая с дескрипторами вроде @property).
"""

print()
```

```python
# 1. Собственное исключение — наследник Exception
class WorkoutNotFoundError(Exception):
    """Тренировка не найдена в расписании клуба."""
    pass

class FitnessClub:
    def __init__(self):
        # Расписание: имя тренировки → список записавшихся
        self.schedule = {
            "Йога": [],
            "Кроссфит": [],
            "Пилатес": [],
        }

    def book_workout(self, member, workout_name):
        if workout_name not in self.schedule:
            raise WorkoutNotFoundError(
                f"Тренировка «{workout_name}» не найдена в расписании клуба. "
                f"Доступные: {', '.join(self.schedule)}"
            )
        self.schedule[workout_name].append(member)
        print(f"Клиент {member} записан на «{workout_name}»")

    def show_schedule(self):
        for name, members in self.schedule.items():
            print(f"  {name}: {len(members)} чел.")

club = FitnessClub()

# Успешная запись
club.book_workout("Иван Петров", "Йога")
club.book_workout("Мария Сидорова", "Кроссфит")

# Попытка записаться на несуществующую тренировку
try:
    club.book_workout("Пётр Иванов", "Бокс")
except WorkoutNotFoundError as e:
    print(f"Ошибка записи: {e}")

# Убедимся, что клуб продолжил работать
club.book_workout("Ольга Смирнова", "Пилатес")
print("\nРасписание:")
club.show_schedule()
"""
ЗАЧЕМ СОЗДАВАТЬ СОБСТВЕННЫЕ ИСКЛЮЧЕНИЯ ВМЕСТО СТАНДАРТНЫХ:

1. ТОЧНОСТЬ СМЫСЛА.
   Стандартное ValueError("Тренировка не найдена") технически сработает,
   но оно не говорит, ЧТО именно пошло не так. Собственное
   WorkoutNotFoundError сразу кричит: «дело в тренировке, а не в чём-то ещё».

5. ИЕРАРХИЯ ИСКЛЮЧЕНИЙ.
   Собственные исключения удобно выстраивать в иерархию:

       class FitnessClubError(Exception): ...
       class WorkoutNotFoundError(FitnessClubError): ...
       class MemberAlreadyBookedError(FitnessClubError): ...
       class ClubIsFullError(FitnessClubError): ...

   Тогда можно ловить и точечно:

       except WorkoutNotFoundError: ...   # только эта

   и широко, всё сразу:

       except FitnessClubError: ...       # любая ошибка клуба

ПРАВИЛА ХОРОШЕГО ТОНА:

- Наследуйся от Exception (или от более общего своего исключения),
  НЕ от BaseException. BaseException — для системных вещей (KeyboardInterrupt,
  SystemExit), их ловить обычным except нельзя.
- Название — существительное с суффиксом Error: WorkoutNotFoundError,
  InvalidMemberError, ClubIsFullError.
- Давай понятное сообщение с деталями: что искали, что доступно.
- Если нужно, добавляй поля для программной обработки (e.workout_name).
- Группируй свои исключения в общий базовый класс — удобно ловить одним except.

КОГДА НУЖНО СВОЁ:

- Ошибка специфична для предметной области (нет тренировки, нет клиента,
  зал переполнен).
- Её должен ловить вызывающий код и реагировать по-разному.
- Хочется добавить дополнительные данные (имя тренировки, список доступных).
- Строишь библиотеку/модуль со своим API.
- Нужна иерархия ошибок: «все ошибки клуба», «все ошибки записи» и т.п.
"""

print()
```

```python
class FitnessClub:
    def __init__(self, name):
        self.name = name
        self.workouts = []   # клуб ХРАНИТ тренировки — они принадлежат ему

    def add_workout(self, workout):
        self.workouts.append(workout)

    def show_workouts(self):
        print(f"Клуб «{self.name}» — тренировки:")
        for w in self.workouts:
            trainer_name = w.trainer.name if w.trainer else "без тренера"
            print(f"  - {w.name} (тренер: {trainer_name})")

    def __repr__(self):
        return f"FitnessClub({self.name!r})"


class Workout:
    def __init__(self, name, club):
        self.name = name
        self.club = club         # АССОЦИАЦИЯ: ссылка на клуб (не владение)
        self.trainer = None      # АССОЦИАЦИЯ: ссылка на тренера (может быть None)
        club.add_workout(self)   # регистрируем тренировку в клубе

    def __repr__(self):
        return f"Workout({self.name!r}, club={self.club.name!r})"


class Trainer:
    def __init__(self, name):
        self.name = name
        self.workouts = []       # АГРЕГАЦИЯ: тренер ЗНАЕТ о тренировках,
                                 # но НЕ владеет ими

    def add_workout(self, workout):
        self.workouts.append(workout)
        workout.trainer = self   # обратная ссылка — двусторонняя ассоциация

    def show_workouts(self):
        print(f"Тренер {self.name} ведёт:")
        for w in self.workouts:
            print(f"  - {w.name} (клуб: {w.club.name})")

    def __repr__(self):
        return f"Trainer({self.name!r})"

# 1. Создаём клуб (он существует сам по себе)
club = FitnessClub("СпортЛайф")

# 2. Создаём тренировки — они принадлежат КЛУБУ
yoga = Workout("Йога", club)
crossfit = Workout("Кроссфит", club)
pilates = Workout("Пилатес", club)

# 3. Создаём тренера — он существует отдельно
trainer = Trainer("Алексей")

# 4. Связываем: тренер ВЕДЁТ тренировки (агрегация)
trainer.add_workout(yoga)
trainer.add_workout(crossfit)

print("=== До удаления тренера ===")
trainer.show_workouts()
club.show_workouts()

# 5. Удаляем тренера
print("\n=== Удаляем тренера ===")
del trainer

# 6. Тренировки ОСТАЛИСЬ в клубе — агрегация в действии
print("\n=== После удаления тренера ===")
club.show_workouts()

"""
ВЫВОД ПРОГРАММЫ:

=== До удаления тренера ===
Тренер Алексей ведёт:
  - Йога (клуб: СпортЛайф)
  - Кроссфит (клуб: СпортЛайф)
Клуб «СпортЛайф» — тренировки:
  - Йога (тренер: Алексей)
  - Кроссфит (тренер: Алексей)
  - Пилатес (тренер: без тренера)

=== Удаляем тренера ===

=== После удаления тренера ===
Клуб «СпортЛайф» — тренировки:
  - Йога (тренер: без тренера)
  - Кроссфит (тренер: без тренера)
  - Пилатес (тренер: без тренера)
"""

"""
ТРИ ВИДА СВЯЗЕЙ МЕЖДУ ОБЪЕКТАМИ:

┌────────────────┬──────────────────────────┬──────────────────────────────┐
│ Вид            │ Что это                  │ Кто кого создаёт / владеет   │
├────────────────┼──────────────────────────┼──────────────────────────────┤
│ КОМПОЗИЦИЯ     │ Жёсткая связь: «часть»   │ Целое СОЗДАЁТ часть внутри   │
│                │ живёт только внутри      │ себя. Часть умирает вместе   │
│                │ «целого»                 │ с целым.
        │
│ АГРЕГАЦИЯ      │ Мягкая связь: «целое»    │ Объекты создаются НЕЗАВИСИМО.│
│                │ ХРАНИТ ссылки на части,  │ Целое лишь ссылается на них. │
│                │ но не владеет ими        │ Части живут сами по себе.
    │
│ АССОЦИАЦИЯ     │ Просто ссылка между      │ Объекты полностью независимы.│
│                │ независимыми объектами   │ Ссылка — «знает о», а не     │
│                │                          │ «владеет».                   │
└────────────────┴──────────────────────────┴──────────────────────────────┘

КОМПОЗИЦИЯ (пример — задача 8, Member + Subscription):
    class Member:
        def __init__(self, name):
            self.subscription = Subscription()   # ← создаётся ВНУТРИ
- Абонемент не существует без клиента.
- Удалили клиента — абонемент тоже исчез (на него больше нет ссылок).
- Жизненный цикл части полностью подчинён целому.

АГРЕГАЦИЯ (наш Trainer + Workout):
    trainer.add_workout(yoga)   # ← тренировка создана СНАРУЖИ
- Тренировка существует независимо от тренера.
- Одна тренировка может перейти к другому тренеру.
- Тренер лишь ХРАНИТ СПИСОК ссылок на тренировки.
- Удалили тренера — тренировки продолжают жить в клубе.
- Ключевой признак: объект-часть передаётся в целое ИЗВНЕ.

АССОЦИАЦИЯ (наш Workout → FitnessClub и Workout → Trainer):
    class Workout:
        def __init__(self, name, club):
            self.club = club        # ← просто ссылка
- Тренировка «знает» о клубе, но не владеет им.
- Клуб «знает» о тренировке (через свой список), но ссылка идёт «вбок».
- Оба объекта независимы: удаление любого не ломает другой.
- Ассоциация может быть:
  - однонаправленной (знает только одна сторона),
  - двунаправленной (обе стороны ссылаются друг на друга),
  - 1-к-1, 1-ко-многим, многие-ко-многим.

"""

print()
```

```python
class Member:
    def get_role(self):
        return "Клиент"

class Employee:
    def get_role(self):
        return "Сотрудник"

class TrainerMember(Member, Employee):
    """Человек, который одновременно и клиент, и сотрудник."""
    pass

# Создаём объект
tm = TrainerMember()

# 1. Какой метод вызовется?
print(tm.get_role())            # Клиент

# 2. Смотрим MRO — порядок разрешения методов
print(TrainerMember.__mro__)
# (<class '__main__.TrainerMember'>,
#  <class '__main__.Member'>,
#  <class '__main__.Employee'>,
#  <class 'object'>)

# 3. Демонстрация: если поменять порядок родителей
class TrainerMember2(Employee, Member):
    pass

print(TrainerMember2().get_role())      # Сотрудник
print(TrainerMember2.__mro__)
# TrainerMember2 → Employee → Member → object

"""
ЧТО ТАКОЕ MRO (Method Resolution Order):

MRO — это упорядоченный список классов, в котором Python ищет
атрибут/метод при обращении obj.method(). Строится он в момент
создания класса и хранится в атрибуте __mro__ (или cls.mro()).

Правила:
1. Сначала сам класс.
2. Потом родители — СЛЕВА НАПРАВО в порядке объявления (для простого случая).
3. Потом родители родителей — но с учётом линеаризации C3.
4. В конце — всегда object (корень всех классов).

ЧТО ТАКОЕ «ПРОБЛЕМА РОМБА» (DIAMOND PROBLEM):

Классическая ситуация в языках с множественным наследованием:

           A
          / \
         B   C
          \ /
           D

Класс D наследуется от B и C, а те — оба от A. Возникает вопрос:
если A определяет метод foo(), а B и C его переопределяют,
какую версию унаследует D? И не вызовется ли A.foo() дважды?

В Python проблема ромба решена через C3-линеаризацию:
    MRO(D) = [D, B, C, A, object]
"""

print()
```

```python
from dataclasses import dataclass, field
from datetime import datetime

# === МИКСИН ===
class LoggableMixin:
    """Примесь: даёт методу log() возможность писать в лог.

    Миксин НЕ предназначен для самостоятельного использования:
    у него нет __init__, нет состояния, только узкая функциональность.
    """
    def log(self):
        # self.__class__.__name__ — имя фактического класса объекта,
        # а не миксина. Это важно: работать будет для любого наследника.
        print(f"[{datetime.now():%Y-%m-%d %H:%M:%S}] "
              f"{self.__class__.__name__}: {self}")


# === ДАТАКЛАСС + МИКСИН ===
# Миксин идёт ПЕРВЫМ в списке родителей.
# Это классический приём: методы миксина получают приоритет в MRO.
@dataclass
class Member(LoggableMixin):
    name: str
    subscription_type: str
    months: int = 12

    def is_active(self):
        return self.months > 0


# === СРАВНЕНИЕ С ВЕРСИЕЙ БЕЗ @dataclass ===
#
# БЫЛО (вручную):
#
#     class Member:
#         def __init__(self, name, subscription_type, months=12):
#             self.name = name
#             self.subscription_type = subscription_type
#             self.months = months
#
#         def __repr__(self):
#             return (f"Member(name={self.name!r}, "
#                     f"subscription_type={self.subscription_type!r}, "
#                     f"months={self.months!r})")
#
#         def __eq__(self, other):
#             if not isinstance(other, Member):
#                 return NotImplemented
#             return (self.name == other.name
#                     and self.subscription_type == other.subscription_type
#                     and self.months == other.months)
#
#         def is_active(self):
#             return self.months > 0
#
# СТАЛО (через @dataclass):
#
#     @dataclass
#     class Member(LoggableMixin):
#         name: str
#         subscription_type: str
#         months: int = 12
#
#         def is_active(self):
#             return self.months > 0
#
# @dataclass САМ сгенерировал:
#   - __init__(self, name, subscription_type, months=12)
#   - __repr__ вида "Member(name='Иван', subscription_type='Премиум', months=12)"
#   - __eq__ по всем полям


# === ДЕМОНСТРАЦИЯ ===
m1 = Member("Иван Петров", "Премиум", 12)
m2 = Member("Иван Петров", "Премиум", 12)
m3 = Member("Мария Сидорова", "Стандарт")

# 1. Автоматический __repr__
print(m1)
# Member(name='Иван Петров', subscription_type='Премиум', months=12)

# 2. Автоматический __eq__
print(m1 == m2)        # True  — все поля совпадают
print(m1 == m3)        # False

# 3. Обычные методы работают как раньше
print(m1.is_active())  # True

# 4. Миксин в деле — log() доступен любому Member
m1.log()
m3.log()

"""
ЧТО ДАЁТ @dataclass:

1. АВТОМАТИЧЕСКИЙ __init__ — по аннотациям полей.
       @dataclass
       class Member:
           name: str
           subscription_type: str
           months: int = 12
   Эквивалентно ручному __init__(self, name, subscription_type, months=12).

2. АВТОМАТИЧЕСКИЙ __repr__ — читаемый, с именами полей.

3. АВТОМАТИЧЕСКИЙ __eq__ — по ВСЕМ полям (а не только по двум, как мы
   делали в задаче 12 вручную).

4. ВОЗМОЖНОСТЬ ТОНКОЙ НАСТРОЙКИ:
   - field(default_factory=list) — для изменяемых значений по умолчанию;
   - field(compare=False) — исключить поле из __eq__;
   - field(repr=False) — не показывать поле в __repr__;
   - @dataclass(frozen=True) — сделать объект неизменяемым (появится __hash__);
   - @dataclass(order=True) — добавить <, <=, >, >= (по полям по порядку).

5. ВАЖНОЕ ОТЛИЧИЕ ОТ РУЧНОЙ ВЕРСИИ:
   dataclass по умолчанию НЕ генерирует __hash__, если eq=True и frozen=False.
   Тогда __hash__ = None, и объект нельзя положить в set/dict.
   Чтобы вернуть хэшируемость — либо frozen=True, либо задать __hash__ вручную.

ЧТО ТАКОЕ МИКСИН (MIXIN):

Миксин — это класс, который:
1. НЕ предназначен для самостоятельного использования.
   Сам по себе он не имеет смысла: LoggableMixin() — бессмыслица.
2. Содержит ОДНУ узкую функциональность (логирование, сериализацию,
   сравнение, кеширование и т.п.).
3. Обычно НЕ имеет собственного состояния (нет полей в __init__).
4. Подключается через множественное наследование, чтобы «подмешать»
   своё поведение в целевой класс.
5. В списке родителей идёт ПЕРВЫМ — тогда его методы получают
   приоритет в MRO и не перекрываются методами основного класса.

Классические примеры миксинов в стандартной библиотеке:
- collections.abc — миксины для протоколов (Iterable, Sequence и т.п.);
- socketserver.ThreadingMixIn — добавляет многопоточность серверу;
- logging — примеси для добавления логирования.

ОТЛИЧИЕ МИКСИНА ОТ ОБЫЧНОГО РОДИТЕЛЬСКОГО КЛАССА:

┌────────────────────┬───────────────────────────┬────────────────────────────┐
│ Критерий           │ Обычный родитель          │ Миксин                     │
├────────────────────┼───────────────────────────┼────────────────────────────┤
│ Смысл отдельно     │ Самостоятелен             │ Бессмысленен в одиночку    │
│ Состояние          │ Может иметь свои поля     │ Обычно БЕЗ состояния       │
│ Объём              │ Целостная сущность        │ Одна узкая функциональность│
│ Иерархия           │ «Является» (is-a)         │ «Умеет» (can-do)           │
│ Позиция в скобках  │ Обычно один или главный   │ Первым в списке родителей  │
│ Пример             │ class Dog(Animal)         │ class Dog(Loggable, Animal)│
└────────────────────┴───────────────────────────┴────────────────────────────┘

КЛЮЧЕВАЯ ИДЕЯ:
Обычный родитель = «ЯВЛЯЕТСЯ» (Member — это человек, клиент).
Миксин            = «УМЕЕТ» (Member — умеет логироваться).

ПОЧЕМУ МИКСИН СТАВЯТ ПЕРВЫМ:

Порядок в class Member(LoggableMixin): определяется MRO:

    Member → LoggableMixin → object

При вызове m.log() Python ищет log() в MRO:
    Member          — нет log()
    LoggableMixin   — ЕСТЬ log()  ← нашли, вызываем.

Если бы миксин стоял ВТОРЫМ (class Member(SomeBase, LoggableMixin)),
а у SomeBase тоже был бы метод log() — выиграл бы SomeBase, и наш миксин
был бы проигнорирован. Поэтому «примеси» ставят первыми — чтобы их
узкая функциональность гарантированно «подмешалась» к классу.

ДОПОЛНИТЕЛЬНО: почему в log() используется self.__class__.__name__, а не
LoggableMixin.__name__:

Если написать жёстко:
    print(f"LoggableMixin: ...")   # ← всегда "LoggableMixin", плохо

то для любого наследника в логе будет одно и то же имя — теряется смысл.
А self.__class__.__name__ даёт ФАКТИЧЕСКИЙ класс объекта:
    m1.log()   → "Member"
    e.log()    → "Employee"     (если Employee тоже с миксином)
    tm.log()   → "TrainerMember" (наследник)

Миксин становится универсальным: одна и та же примесь корректно работает
для всех, кто её подключил.

КОГДА СТОИТ ДЕЛАТЬ МИКСИН:

- Функциональность повторяется в разных, не связанных родством классах
  (логирование, сериализация, сравнение, кеш).
- Не хочется плодить глубокую иерархию наследования ради одной функции.
- Логика не зависит от внутреннего устройства класса — работает через
  общий интерфейс (например, через __str__ или через поля, которые
  гарантированно есть у целевого класса).

КОГДА МИКСИН — ПЛОХАЯ ИДЕЯ:

- Если примесь требует много полей в __init__ — это уже не миксин,
  а полноценный родитель.
- Если примесь рассчитана только на один конкретный класс — проще
  добавить метод прямо туда.
- Если примесей становится много и они конфликтуют — MRO превращается
  в кашу, отлаживать больно.

ИТОГ:
- @dataclass избавляет от рутинного __init__/__repr__/__eq__ — код
  становится короче и читаемее.
- Миксин — это «примесь» с одной функциональностью, без своего
  состояния, для подключения через множественное наследование.
- Миксин ставят первым в списке родителей, чтобы его методы
  выигрывали по MRO.
- Обычный родитель говорит «это ЯВЛЯЕТСЯ», миксин — «это УМЕЕТ».
"""

print()
```

```python
import sys

# === ВЕРСИЯ БЕЗ __slots__ ===
class MemberNoSlots:
    def __init__(self, name, membership_type, months):
        self.name = name
        self.membership_type = membership_type
        self.months = months

# === ВЕРСИЯ С __slots__ ===
class Member:
    __slots__ = ("name", "membership_type", "months")

    def __init__(self, name, membership_type, months):
        self.name = name
        self.membership_type = membership_type
        self.months = months


# 1. Создаём объекты
m_normal = MemberNoSlots("Иван Петров", "Премиум", 12)
m_slots = Member("Иван Петров", "Премиум", 12)

# 2. Размер «оболочки» через sys.getsizeof
print("sys.getsizeof (только оболочка объекта):")
print(f"  без __slots__: {sys.getsizeof(m_normal)} байт")
print(f"  с __slots__:   {sys.getsizeof(m_slots)} байт")

# 3. Реальная память с учётом __dict__ и содержимого
#    (без pympler, вручную — через __dict__ и sys.getsizeof полей)
print("\nРучная оценка:")
print(f"  m_normal.__dict__: {m_normal.__dict__}")
print(f"  размер __dict__:  {sys.getsizeof(m_normal.__dict__)} байт")
print(f"  m_slots: нет __dict__ → " +
      ("есть" if hasattr(m_slots, "__dict__") else "нет"))

# 4. Суммарная оценка на 1 объект (приблизительно):
size_normal = sys.getsizeof(m_normal) + sys.getsizeof(m_normal.__dict__)
size_slots = sys.getsizeof(m_slots)
print(f"\nПриблизительный размер объекта:")
print(f"  без __slots__: ~{size_normal} байт")
print(f"  с __slots__:   ~{size_slots} байт")
print(f"  экономия:      ~{size_normal - size_slots} байт на объект")

# 5. При 100 000 объектах
N = 100_000
print(f"\nПри {N:,} объектов:")
print(f"  без __slots__: ~{(size_normal * N) / 1024 / 1024:.1f} МБ")
print(f"  с __slots__:   ~{(size_slots * N) / 1024 / 1024:.1f} МБ")

# 6. Нельзя добавить новый атрибут
try:
    m_slots.phone = "+7-999-123-45-67"
except AttributeError as e:
    print(f"\nAttributeError: {e}")

# А без __slots__ — можно
m_normal.phone = "+7-999-123-45-67"
print(f"m_normal.phone = {m_normal.phone}")   # работает

# 7. Существующие поля читаются/пишутся нормально
m_slots.months = 24
print(f"\nm_slots.months = {m_slots.months}")

"""
ВЫВОД ПРОГРАММЫ (числа могут немного отличаться по версии Python):

sys.getsizeof (только оболочка объекта):
  без __slots__: 48 байт
  с __slots__:   56 байт

Ручная оценка:
  m_normal.__dict__: {'name': 'Иван Петров', 'membership_type': 'Премиум', 'months': 12}
  размер __dict__:  296 байт
  m_slots: нет __dict__ → нет

Приблизительный размер объекта:
  без __slots__: ~344 байт
  с __slots__:   ~56 байт
  экономия:      ~288 байт на объект

При 100 000 объектов:
  без __slots__: ~32.8 МБ
  с __slots__:   ~5.3 МБ

AttributeError: 'Member' object has no attribute 'phone'
m_normal.phone = +7-999-123-45-67

m_slots.months = 24
"""

"""
ВАЖНОЕ ЗАМЕЧАНИЕ ПРО sys.getsizeof:

sys.getsizeof(m_slots) показывает БОЛЬШЕ байт, чем m_normal — и это
НЕ значит, что __slots__ «хуже». Просто getsizeof измеряет только
«оболочку» объекта, не заглядывая внутрь. У объекта с __slots__
оболочка чуть больше (там лежат слоты), но зато НЕТ отдельного
словаря __dict__, который у обычного объекта весит ~184 байта.

Честная оценка — сумма:
    размер объекта + размер __dict__

И вот тут __slots__ выигрывает в разы: ~56 байт против ~232 байт.
На 100 000 объектах это ~17 МБ экономии — существенно.

Для более точного измерения используют сторонние библиотеки:
    pip install pympler
    from pympler import asizeof
    print(asizeof.asizeof(m_normal))   # → например, 232
    print(asizeof.asizeof(m_slots))    # → например, 56

КАК __slots__ ЭКОНОМИТ ПАМЯТЬ:

Обычный объект в Python устроен так:

    ┌──────────────────────────┐
    │ PyObject_HEAD            │  ← служебные поля (refcount, тип)
    ├──────────────────────────┤
    │ __dict__ (указатель) ────┼──→  ┌───────────────────────────┐
    └──────────────────────────┘     │ СЛОВАРЬ атрибутов         │
                                     │ {'name': ..., 'months': …}│
                                     │ (хэш-таблица, ~184+ байт) │
                                     └───────────────────────────┘

С __slots__:

    ┌──────────────────────────┐
    │ PyObject_HEAD            │
    ├──────────────────────────┤
    │ name         ────────────┼──→  "Иван Петров"
    │ membership_type ─────────┼──→  "Премиум"
    │ months       ────────────┼──→  12
    └──────────────────────────┘
    (никакого __dict__ — значения лежат прямо в объекте)

Что происходит:
1. Python НЕ создаёт __dict__ для каждого объекта.
2. Вместо этого под каждый атрибут из __slots__ резервируется
   фиксированное место (слот) прямо в структуре объекта.
3. Доступ к атрибуту идёт напрямую по индексу слота — это ещё и
   БЫСТРЕЕ, чем поиск в словаре.
4. Экономия — в первую очередь за счёт отсутствия __dict__
   (у каждого объекта ~184+ байт).

ЧТО ТЕРЯЕТСЯ ПРИ ИСПОЛЬЗОВАНИИ __slots__:

1. НЕТ __dict__ → нельзя добавлять новые атрибуты «на лету»:
       m.phone = "..."   # AttributeError
   Только те атрибуты, что объявлены в __slots__.

2. НЕЛЬЗЯ использовать @property с тем же именем, что и слот.
   Точнее — нельзя, чтобы имя слота и имя property совпадали:

       class Member:
           __slots__ = ("_months",)

           @property
           def months(self):        # OK, имя "_months" не конфликтует
               return self._months

   А вот так — ошибка:

       class Member:
           __slots__ = ("months",)

           @property
           def months(self):        # ValueError: 'months' in __slots__
               return 0             # conflicts with class variable

   Правило: в __slots__ пиши «служебное» имя (_months),
   а @property — публичное (months).

3. НАСЛЕДОВАНИЕ:
   - Если дочерний класс НЕ определяет свои __slots__, он автоматически
     получает __dict__, и вся экономия теряется:

         class Member:
             __slots__ = ("name", "months")

         class VIPMember(Member):    # ← НЕТ __slots__
             pass

         vip = VIPMember()
         vip.phone = "..."   # работает! У VIPMember снова есть __dict__

   - Если дочерний класс определяет __slots__, он должен ПЕРЕЧИСЛИТЬ
     только СВОИ новые поля. Поля родителя уже «заняты»:

         class VIPMember(Member):
             __slots__ = ("discount",)   # только новое поле

         vip = VIPMember()
         vip.phone = "..."   # AttributeError

4. НЕЛЬЗЯ МНОЖЕСТВЕННО НАСЛЕДОВАТЬ от двух классов с НЕПУСТЫМИ
   __slots__ одновременно — Python не знает, как расположить их
   слоты в памяти и бросит TypeError. Миксины обычно делают
   БЕЗ __slots__ или с __slots__ = ().

5. НЕЛЬЗЯ ИСПОЛЬЗОВАТЬ ВМЕСТЕ С @dataclass БЕЗ ОСТОРОЖНОСТИ.
   @dataclass можно сочетать с __slots__ (Python 3.10+):

       @dataclass(slots=True)
       class Member:
           name: str
           months: int = 12

   В старых версиях — приходилось прописывать __slots__ вручную,
   и там были тонкости с полями по умолчанию.

6. ПРОБЛЕМЫ С PICKLE:
   До Python 3.8 объекты со __slots__ плохо пиклились. Сейчас
   всё работает, но иногда нужно добавить __getstate__/__setstate__.

7. НЕЛЬЗЯ ВЫБИРАТЬ __dict__ ОБРАТНО В РАНТАЙМЕ:
   Есть обходной путь — добавить "__dict__" в __slots__, тогда
   объект снова обзаведётся словарём, но экономия исчезнет.

КОГДА СТОИТ ПРИМЕНЯТЬ __slots__:

✅ Стоит:
- ОЧЕНЬ много объектов (десятки/сотни тысяч, миллионы):
  клиенты, транзакции, точки, записи логов, узлы графа.
- Класс-«запись» с фиксированным набором полей, без «на лету».
- Хочется ускорить доступ к атрибутам (слот быстрее словаря).
- Объекты неизменяемы или почти неизменяемы по структуре.

❌ Не стоит:
- Мало объектов — выигрыш в памяти незаметен, а ограничений много.
- Нужна гибкость: динамические атрибуты, плагины, метаданные.
- Активно используются @property или миксины — придётся аккуратничать.
- Класс предполагает глубокое/множественное наследование — легко
  наступить на грабли (например, забыть __slots__ в наследнике).
- Раньше времени — сначала измерь, потом оптимизируй.

ЗОЛОТОЕ ПРАВИЛО:
__slots__ — это оптимизация. Применяй её, когда есть измеренная
проблема с памятью, а не «на всякий случай». Для 99% классов
обычный __dict__ удобнее и гибче.

ИТОГ:
- __slots__ убирает __dict__ у каждого объекта и хранит значения
  в фиксированных слотах → экономия памяти (до нескольких раз)
  и ускорение доступа.
- Цена: нельзя добавлять новые атрибуты, есть тонкости с @property,
  наследованием и миксинами.
- Использовать — для больших коллекций простых объектов с
  фиксированным набором полей.
- Сочетается с @dataclass: @dataclass(slots=True) (Python 3.10+).

ОТДЕЛЬНО ПРО ПОТЕРЮ __dict__:

С точки зрения Python, наличие __dict__ у объекта означает
«динамическая структура»: можно дописывать что угодно.

    m_normal.__dict__['phone'] = "..."   # обходной путь, работает
    m_normal.phone = "..."                # обычная запись, работает

А у объекта со __slots__ ни __dict__, ни возможности его создать нет:

    m_slots.__dict__                     # AttributeError
    m_slots.phone = "..."                # AttributeError

Именно поэтому __slots__ иногда называют «статической типизацией
атрибутов»: набор полей фиксируется в момент определения класса,
как колонки в таблице БД.
"""

print()
```

```python
class RangeDescriptor:
    """Дескриптор: проверяет, что значение лежит в диапазоне [min, max].

    Хранит значение в instance.__dict__ — то есть у КАЖДОГО объекта
    своё значение. Без этого все объекты делили бы одно общее.
    """

    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value
        self.name = None   # имя атрибута узнаем из __set_name__

    def __set_name__(self, owner, name):
        # Вызывается автоматически при создании класса.
        # owner — сам класс (Member), name — имя атрибута ("months"/"visits").
        # Запоминаем имя, чтобы потом знать, под каким ключом класть
        # значение в instance.__dict__.
        self.name = name

    def __get__(self, instance, owner=None):
        # instance — объект, owner — класс.
        # Если обращаются через КЛАСС (Member.months) — instance is None.
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        # Проверяем тип и диапазон
        if not isinstance(value, int) or isinstance(value, bool):
            raise ValueError(
                f"{self.name}: должно быть целым числом, "
                f"получено {type(value).__name__}: {value!r}"
            )
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(
                f"{self.name}: значение должно быть в диапазоне "
                f"[{self.min_value}, {self.max_value}], получено: {value}"
            )
        # Ключевая строка: храним значение В САМОМ ЭКЗЕМПЛЯРЕ,
        # а не в дескрипторе. Иначе все объекты делили бы одно значение.
        instance.__dict__[self.name] = value

    def __delete__(self, instance):
        del instance.__dict__[self.name]


class Member:
    # Один и тот же дескриптор переиспользуется в разных полях.
    # RangeDescriptor — как «шаблон» с проверкой.
    months = RangeDescriptor(1, 60)
    visits = RangeDescriptor(0, 1000)

    def __init__(self, name, months, visits):
        self.name = name
        self.months = months      # ← вызовет RangeDescriptor.__set__
        self.visits = visits      # ← тоже

    def __repr__(self):
        return (f"Member({self.name!r}, "
                f"months={self.months}, visits={self.visits})")


# === Проверка ===

m = Member("Иван Петров", 12, 100)
print(m)                       # Member('Иван Петров', months=12, visits=100)

# Корректные изменения
m.months = 24
m.visits = 500
print(m)                       # Member('Иван Петров', months=24, visits=500)

# Выход за диапазон
for attr, bad in [("months", 0), ("months", 61),
                  ("visits", -1), ("visits", 1001)]:
    try:
        setattr(m, attr, bad)
        print(f"{attr}={bad} → прошло (не должно было!)")
    except ValueError as e:
        print(f"{attr}={bad} → ValueError: {e}")

# Неверный тип
for attr, bad in [("months", "12"), ("visits", 3.5), ("months", True)]:
    try:
        setattr(m, attr, bad)
        print(f"{attr}={bad!r} → прошло (не должно было!)")
    except ValueError as e:
        print(f"{attr}={bad!r} → ValueError: {e}")

# Значения не изменились после ошибок
print(m)                       # Member('Иван Петров', months=24, visits=500)

# У КАЖДОГО объекта — своё значение (не общее!)
m2 = Member("Мария Сидорова", 6, 10)
print(m.months, m2.months)     # 24 6
print(m.visits, m2.visits)     # 500 10

# Значения лежат в __dict__ КАЖДОГО экземпляра
print(m.__dict__)              # {'name': 'Иван Петров', 'months': 24, 'visits': 500}
print(m2.__dict__)             # {'name': 'Мария Сидорова', 'months': 6, 'visits': 10}

# Обращение через класс возвращает сам дескриптор
print(Member.months)           # <__main__.RangeDescriptor object at 0x...>

"""
ВЫВОД ПРОГРАММЫ (примерный):

Member('Иван Петров', months=12, visits=100)
Member('Иван Петров', months=24, visits=500)
months=0 → ValueError: months: значение должно быть в диапазоне [1, 60], получено: 0
months=61 → ValueError: months: значение должно быть в диапазоне [1, 60], получено: 61
visits=-1 → ValueError: visits: значение должно быть в диапазоне [0, 1000], получено: -1
visits=1001 → ValueError: visits: значение должно быть в диапазоне [0, 1000], получено: 1001
months='12' → ValueError: months: должно быть целым числом, получено str: '12'
visits=3.5 → ValueError: visits: должно быть целым числом, получено float: 3.5
months=True → ValueError: months: должно быть целым числом, получено bool: True
Member('Иван Петров', months=24, visits=500)
24 6
500 10
{'name': 'Иван Петров', 'months': 24, 'visits': 500}
{'name': 'Мария Сидорова', 'months': 6, 'visits': 10}
<__main__.RangeDescriptor object at 0x7f8b1c...>
"""

"""
КАК ДЕСКРИПТОРЫ РАБОТАЮТ «ПОД КАПОТОМ» @property:

Дескриптор — это объект класса, у которого определены методы
__get__, __set__ или __delete__. Если такой объект лежит
атрибутом КЛАССА, Python перехватывает обращения к нему у экземпляра
и вызывает соответствующий dunder-метод.

    class Member:
        months = RangeDescriptor(1, 60)   # ← дескриптор лежит в классе

    m = Member(...)
    m.months             # → type(m).__dict__['months'].__get__(m, Member)
    m.months = 12        # → type(m).__dict__['months'].__set__(m, 12)
    del m.months         # → type(m).__dict__['months'].__delete__(m)

Правила приоритета при чтении obj.attr:
1. data-дескриптор (есть __get__ И __set__/__delete__) у КЛАССА — побеждает.
2. Запись в obj.__dict__.
3. non-data-дескриптор (только __get__) у класса.
4. Атрибут класса.
5. __getattr__ (если определён).
6. AttributeError.

@property — это ГОТОВЫЙ дескриптор из стандартной библиотеки:

    class property:
        def __init__(self, fget=None, fset=None, fdel=None, doc=None):
            self.fget = fget
            self.fset = fset
            ...
        def __get__(self, instance, owner=None):
            return self.fget(instance)
        def __set__(self, instance, value):
            self.fset(instance, value)
        def __delete__(self, instance):
            self.fdel(instance)

То есть когда ты пишешь:

    class Member:
        @property
        def months(self):
            return self._months

        @months.setter
        def months(self, value):
            ...

Python создаёт объект property, кладёт его в Member.__dict__['months'],
и дальше всё работает ровно так же, как с нашим RangeDescriptor.
@property — это «дескриптор для одного поля», а RangeDescriptor —
«дескриптор-шаблон для многих полей с одной проверкой».

ЗАЧЕМ НУЖЕН __set_name__:

До Python 3.6 дескриптор не знал, под каким именем его положили в класс.
Приходилось либо передавать имя вручную:

    months = RangeDescriptor(1, 60, name="months")   # ← костыль

либо копаться в __dict__ класса:

    self.name = [k for k, v in vars(owner).items() if v is self][0]

Python 3.6 добавил протокол __set_name__: когда класс создаётся,
Python САМ вызывает __set_name__ у каждого дескриптора, лежащего
в теле класса, и передаёт ему owner и имя:

    class Member:
        months = RangeDescriptor(1, 60)   # → months.__set_name__(Member, "months")
        visits = RangeDescriptor(0, 1000) # → visits.__set_name__(Member, "visits")

Внутри мы сохраняем имя:

    def __set_name__(self, owner, name):
        self.name = name

Зачем именно имя? Чтобы знать, под каким ключом класть значение
в instance.__dict__:

    instance.__dict__[self.name] = value   # {"months": 12} вместо {"?": 12}

Без __set_name__ дескриптор не знал бы, какое имя использовать,
и не смог бы хранить значения в экземпляре корректно.

ПОЧЕМУ ВАЖНО ХРАНИТЬ ЗНАЧЕНИЕ В instance.__dict__, А НЕ В ДЕСКРИПТОРЕ:

Допустим, мы ошиблись и написали:

    def __set__(self, instance, value):
        ...
        self.value = value    # ← ПЛОХО: значение в дескрипторе!

Что произойдёт:

    months = RangeDescriptor(1, 60)   # ОДИН объект-дескриптор на ВЕСЬ класс
    m1 = Member(...); m1.months = 12  # self.value = 12
    m2 = Member(...); m2.months = 6   # self.value = 6   ← перезаписали!
    print(m1.months)                  # 6 — но должно быть 12! ❌

Все экземпляры делят ОДИН объект-дескриптор, потому что он лежит
в классе, а не в экземпляре. Значит, хранить в нём данные экземпляра
нельзя — они будут общими.

Правильный подход:

    def __set__(self, instance, value):
        ...
        instance.__dict__[self.name] = value

Тогда у КАЖДОГО экземпляра своя запись:

    m1.__dict__ = {"name": "Иван", "months": 12, "visits": 100}
    m2.__dict__ = {"name": "Мария", "months": 6,  "visits": 10}

Дескриптор — «логика», данные — в экземпляре. Это ключевая идея
всех data-дескрипторов.

ПРЕИМУЩЕСТВО ДЕСКРИПТОРА ПЕРЕД @property:

Представь, что у класса 10 полей с одинаковой проверкой «целое > 0»:

Через @property (много дублирования):

    class Member:
        @property
        def months(self): return self._months
        @months.setter
        def months(self, v):
            if not isinstance(v, int) or v <= 0: raise ValueError(...)
            self._months = v

        @property
        def visits(self): return self._visits
        @visits.setter
        def visits(self, v):
            if not isinstance(v, int) or v <= 0: raise ValueError(...)
            self._visits = v

        @property
        def points(self): return self._points
        @points.setter
        def points(self, v):
            if not isinstance(v, int) or v <= 0: raise ValueError(...)
            self._points = v
        # ... ещё 7 таких же блоков. 30+ строк рутины, копипаста,
        # при изменении правил надо править все 10 мест.

Через дескриптор (один раз описали — используем многократно):

    class Member:
        months = PositiveInt()
        visits = PositiveInt()
        points = PositiveInt()
        # ... ещё 7 полей — по одной строке на каждое.
        # Логика в ОДНОМ месте: класс PositiveInt.

Преимущества дескриптора:
1. DRY (Don't Repeat Yourself): одна проверка — один класс.
2. Переиспользование: один дескриптор — десятки полей.
3. Параметризация: RangeDescriptor(1, 60) и RangeDescriptor(0, 1000) —
   один класс, разные диапазоны.
4. Композиция проверок: можно собрать цепочку дескрипторов.
5. Тестируемость: тестируем дескриптор отдельно, а не каждое property.
6. Меньше кода: 10 полей = 10 строк вместо 10*4 = 40.
7. Единое место правки: меняешь правила валидации в одном классе.

Недостатки дескриптора:
- Сложнее для понимания новичку, чем @property.
- Больше «магии» — отладка чуть тяжелее.
- Для ОДНОГО поля дескриптор — оверинжиниринг, @property проще.

ЗОЛОТОЕ ПРАВИЛО:
- Одно поле, одна проверка → @property.
- Много полей, одинаковая проверка → дескриптор.
- Совсем сложная логика, зависящая от класса → дескриптор + __set_name__.

ИТОГ:
- Дескриптор — объект с __get__/__set__/__delete__, лежащий в классе.
  Python автоматически вызывает его при обращении к атрибуту экземпляра.
- @property — частный случай дескриптора из стандартной библиотеки.
- Значение нужно хранить в instance.__dict__[self.name], иначе все
  экземпляры будут делить одно значение.
- __set_name__ (Python 3.6+) сообщает дескриптору его имя в классе,
  чтобы знать, под каким ключом писать в instance.__dict__.
- Главное преимущество дескриптора — переиспользование одной логики
  для многих полей. Для одного поля удобнее @property.

В СТАНДАРТНОЙ БИБЛИОТЕКЕ дескрипторы используются повсюду:
- @property, @classmethod, @staticmethod — все дескрипторы;
- functions — тоже дескрипторы (поэтому методы работают);
- slots — реализованы через дескрипторы;
- SQLAlchemy Column, Django Field — дескрипторы.
"""

print()
```

```python
class SingletonMeta(type):
    """Метакласс, гарантирующий, что у класса будет только один экземпляр."""

    # Словарь: класс → его единственный экземпляр.
    # Храним в метаклассе, а не в самом классе, чтобы не мешать
    # экземплярным атрибутам.
    _instances = {}

    def __call__(cls, *args, **kwargs):
        # __call__ метакласса вызывается, когда мы делаем SomeClass(...).
        # Здесь перехватываем создание объекта.
        if cls not in cls._instances:
            # Первый раз — создаём через обычный type.__call__,
            # который вызовет __new__ и __init__.
            cls._instances[cls] = super().__call__(*args, **kwargs)
        # Все последующие вызовы возвращают ТОТ ЖЕ объект.
        return cls._instances[cls]


class FitnessClub(metaclass=SingletonMeta):
    def __init__(self, name="СпортЛайф"):
        self.name = name
        self.members = []

    def add_member(self, member):
        self.members.append(member)

    def __repr__(self):
        return f"FitnessClub({self.name!r}, members={len(self.members)})"


# === Проверка ===

club1 = FitnessClub("СпортЛайф")
club2 = FitnessClub("Другой клуб")

print(club1 is club2)         # True — это ОДИН и тот же объект
print(club1.name)             # СпортЛайф — имя от первого вызова
print(club2.name)             # СпортЛайф — второй вызов не создал новый

# Изменения через одну «ссылку» видны в другой
club1.add_member("Иван")
club2.add_member("Мария")
print(club1.members)          # ['Иван', 'Мария']
print(club2.members)          # ['Иван', 'Мария'] — тот же список!
print(len(club1.members))     # 2

# Оба имени указывают на один объект
print(id(club1) == id(club2)) # True

# Что произошло бы без SingletonMeta:
# club1 = FitnessClub("СпортЛайф")
# club2 = FitnessClub("Другой клуб")
# club1 is club2   # False — два разных объекта, данные разошлись бы

"""
ВЫВОД ПРОГРАММЫ:

True
СпортЛайф
СпортЛайф
['Иван', 'Мария']
['Иван', 'Мария']
2
True
"""

"""
ЧТО ТАКОЕ МЕТАКЛАСС:

Обычный класс — это «чертёж» для объектов.
Метакласс — это «чертёж» для классов.

    ┌─────────────────┐   создаёт   ┌──────────────┐   создаёт   ┌──────────┐
    │  Метакласс      │ ─────────→  │   Класс      │ ─────────→  │  Объект  │
    │  SingletonMeta  │             │ FitnessClub  │             │  club1   │
    └─────────────────┘             └──────────────┘             └──────────┘
         сам является                  является
         экземпляром type              экземпляром SingletonMeta

Ключевая мысль: ВСЁ в Python — объект. Классы тоже объекты.
У каждого объекта есть тип (класс). У класса тоже есть тип — это
и есть метакласс.

    type(club1)             # <class 'FitnessClub'>       — тип объекта
    type(FitnessClub)       # <class 'SingletonMeta'>     — тип класса = метакласс
    type(SingletonMeta)     # <class 'type'>              — тип метакласса
    type(type)              # <class 'type'>              — type — сам себе тип

По умолчанию метакласс любого класса — это type:

    class Member: ...        # то же самое, что
    Member = type("Member", (), {})   # ← динамическое создание класса

Когда пишешь class FitnessClub(metaclass=SingletonMeta), Python вместо
type вызывает SingletonMeta для создания класса FitnessClub.

ГДЕ МЕТАКЛАСС «ВСТРАИВАЕТСЯ» В ЖИЗНЕННЫЙ ЦИКЛ КЛАССА:

1. class FitnessClub(metaclass=SingletonMeta): ... — Python начинает
   создавать класс.
2. Вызывается SingletonMeta.__new__(mcs, name, bases, namespace) —
   создаётся сам класс (объект типа SingletonMeta).
3. Вызывается SingletonMeta.__init__(cls, name, bases, namespace) —
   инициализация класса.
4. Позже, когда пишем club = FitnessClub(...) — вызывается
   SingletonMeta.__call__(cls, ...), и вот тут мы перехватываем
   создание экземпляра.

В нашей задаче мы переопределили именно __call__, чтобы контролировать
процесс FitnessClub(...). Если бы мы переопределили __new__ или __init__
метакласса, то управляли бы СОЗДАНИЕМ КЛАССА, а не объекта.

КАК РАБОТАЕТ НАШ SINGLETON:

    _instances = {}    # один словарь на метакласс, общий для всех его классов

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

Пошагово для двух вызовов:

    club1 = FitnessClub("СпортЛайф")
        __call__(FitnessClub, "СпортЛайф")
        FitnessClub нет в _instances
        → super().__call__() → обычное создание → __new__ + __init__
        → _instances[FitnessClub] = <FitnessClub #1>
        → return <FitnessClub #1>

    club2 = FitnessClub("Другой клуб")
        __call__(FitnessClub, "Другой клуб")
        FitnessClub УЖЕ есть в _instances
        → return _instances[FitnessClub]   # <FitnessClub #1> — тот же!
        (аргумент "Другой клуб" просто игнорируется)

Поэтому club1 is club2 → True. __init__ второго раза НЕ вызывается,
значит имя остаётся "СпортЛайф", а список members — общий.

ЗАЧЕМ ВООБЩЕ НУЖНЫ МЕТАКЛАССЫ:

Метаклассы позволяют вмешаться в САМ ПРОЦЕСС СОЗДАНИЯ КЛАССА:
- изменить/дополнить набор методов и атрибутов;
- автоматически регистрировать классы в реестре;
- проверять корректность определения класса (например, что
  у каждого метода есть docstring);
- генерировать код «на лету»;
- реализовать паттерны вроде Singleton, Abstract Base Class, ORM.

Что можно сделать через метакласс, а что — обычными средствами:

| Задача                                  | Метакласс | Обычные средства          |
|-----------------------------------------|-----------|---------------------------|
| Один экземпляр класса                   | ✅ удобно | модуль, Borg, декоратор   |
| Автоматическая регистрация наследников  | ✅        | __init_subclass__ (3.6+)  |
| Проверка правил при создании класса     | ✅        | __init_subclass__         |
| Добавление методов всем классам         | ✅        | декоратор класса          |
| Изменение поведения экземпляров         | ❌        | __getattr__, дескрипторы  |

ПОЧЕМУ МЕТАКЛАССЫ РЕДКО ПИШУТ ВРУЧНУЮ:

В 95% проектов метаклассы НЕ НУЖНЫ. Причины:
1. Слишком мощный инструмент: одно изменение метакласса влияет
   на ВСЕ классы, которые его используют. Легко наломать дров.
2. Трудночитаемо: незнакомый разработчик не сразу понимает,
   откуда у класса взялись методы или атрибуты.
3. Сложно отлаживать: ошибки в метаклассе всплывают при создании
   класса, а не при его использовании — стек вызовов путаный.
4. Часто есть более простые альтернативы:
   - __init_subclass__ (Python 3.6+) — для автоматической настройки
     подклассов;
   - декораторы классов — для изменения класса;
   - @classmethod — для фабричных методов;
   - композиция — вместо наследования от метакласса.

ЗАЧЕМ ЖЕ ИХ ПОНИМАТЬ:

Даже если ты не пишешь метаклассы сам, ты их ВСТРЕТИШЬ в чужом коде:

1. ФРЕЙМВОРКИ:
   - Django: Model использует ModelBase (метакласс) — он читает
     объявление полей и строит внутреннюю структуру модели.
   - SQLAlchemy: DeclarativeMeta — превращает класс с Column-полями
     в маппинг на таблицу БД.
   - Pydantic: ModelMetaclass — собирает схему валидации из аннотаций.
   - abc.ABCMeta — метакласс для абстрактных классов.

2. СТАНДАРТНАЯ БИБЛИОТЕКА:
   - enum.EnumMeta — метакласс для перечислений.
   - type — базовый метакласс для всего.

3. ОТЛАДКА:
   Если видишь странное поведение класса, которое не объясняется
   ни __init__, ни __new__, ни декораторами — возможно, дело
   в метаклассе. Проверить можно так:

       print(type(SomeClass))          # какой метакласс?
       print(SomeClass.__mro__)        # цепочка наследования
       print(vars(SomeClass))          # что реально лежит в классе

4. АРХИТЕКТУРА:
   Понимание метаклассов даёт «полную картину» модели объектов Python:
   объект → класс → метакласс → type. Без этого не получится
   осознанно работать с дескрипторами, __slots__, ORM и DI-контейнерами.

АЛЬТЕРНАТИВЫ SINGLETON В PYTHON:

Метакласс — не единственный способ. Часто проще и понятнее так:

1. Модуль как синглтон (самый «питоничный»):
       # club.py
       club = FitnessClub("СпортЛайф")
   # в других файлах:
       from club import club
   Python гарантирует, что модуль импортируется один раз.

2. @classmethod get_instance():
       class FitnessClub:
           _instance = None
           @classmethod
           def get_instance(cls):
               if cls._instance is None:
                   cls._instance = cls()
               return cls._instance

3. Декоратор @singleton (функциональный аналог метакласса).

4. Паттерн Borg (моносостояние) — все экземпляры делят состояние,
   но остаются разными объектами.

Метакласс оправдан, когда синглтонов МНОГО и хочется описать
правило один раз и «навешивать» на разные классы:

    class Database(metaclass=SingletonMeta): ...
    class Logger(metaclass=SingletonMeta): ...
    class Config(metaclass=SingletonMeta): ...

ВАЖНЫЕ ОГОВОРКИ ПРО SINGLETON В PYTHON:

1. Синглтон мешает тестированию: состояние сохраняется между тестами,
   приходится «сбрасывать» вручную.
2. Синглтон — это глобальное состояние под другим именем; в многопоточном
   коде нужна блокировка (threading.Lock), иначе два потока могут
   создать два экземпляра.
3. Иногда синглтон — «запах» архитектуры: лучше передавать зависимости
   явно (внедрение зависимостей), чем прятать их в глобальный объект.

ИТОГ:
- Метакласс — это класс для классов. Обычный класс создаёт объекты,
  метакласс создаёт классы. По умолчанию метакласс — type.
- Класс — экземпляр своего метакласса: type(FitnessClub) == SingletonMeta.
- Переопределяя __call__ метакласса, можно перехватить создание
  экземпляров и реализовать Singleton.
- Метаклассы — мощный, но редкий инструмент: в 95% проектов их
  не пишут. Но понимать их важно, потому что они лежат в основе
  Django, SQLAlchemy, Pydantic, enum, abc.
- Прежде чем писать метакласс, подумай об альтернативах:
  __init_subclass__, декоратор класса, модуль-синглтон, композиция.

ЗОЛОТОЕ ПРАВИЛО: «Если сомневаешься, нужен ли метакласс — он не нужен».
"""

print()
```

```python
# === СПОСОБ 1: через метакласс ===
class SingletonMeta(type):
    """Метакласс-одиночка: один экземпляр на каждый класс."""

    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]


class MemberRegistryMeta(metaclass=SingletonMeta):
    """Реестр клиентов через метакласс."""

    def __init__(self):
        self.members = []

    def add(self, member):
        self.members.append(member)

    def show(self):
        print(f"Реестр (метакласс): {self.members}")


# === СПОСОБ 2: через __new__ ===
class MemberRegistryNew:
    """Реестр клиентов через переопределение __new__."""

    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            # super().__new__(cls) создаёт «пустой» объект класса cls.
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self):
        # ВАЖНАЯ ЛОВУШКА: __init__ будет вызываться ПРИ КАЖДОМ
        # обращении MemberRegistryNew(), потому что __new__ каждый раз
        # возвращает один и тот же объект, а Python после __new__
        # всегда вызывает __init__ у возвращённого объекта.
        # Поэтому нужна защита «инициализируем только один раз».
        if not hasattr(self, "_initialized"):
            self.members = []
            self._initialized = True

    def add(self, member):
        self.members.append(member)

    def show(self):
        print(f"Реестр (__new__): {self.members}")


# === ПРОВЕРКА СПОСОБА 1 (метакласс) ===
print("=== Метакласс ===")
r1 = MemberRegistryMeta()
r2 = MemberRegistryMeta()

print(r1 is r2)          # True — один и тот же объект
r1.add("Иван")
r2.add("Мария")
r1.show()                # Реестр (метакласс): ['Иван', 'Мария']
r2.show()                # Реестр (метакласс): ['Иван', 'Мария']

# Метакласс «работает автоматически» для наследников
class VipRegistryMeta(MemberRegistryMeta):
    pass

v1 = VipRegistryMeta()
v2 = VipRegistryMeta()
print(v1 is v2)          # True — у каждого класса свой единственный экземпляр
print(v1 is r1)          # False — это РАЗНЫЕ синглтоны (VipRegistry ≠ MemberRegistry)


# === ПРОВЕРКА СПОСОБА 2 (__new__) ===
print("\n=== __new__ ===")
n1 = MemberRegistryNew()
n2 = MemberRegistryNew()

print(n1 is n2)          # True — один и тот же объект
n1.add("Пётр")
n2.add("Ольга")
n1.show()                # Реестр (__new__): ['Пётр', 'Ольга']
n2.show()                # Реестр (__new__): ['Пётр', 'Ольга']

# А что с наследниками у __new__-варианта?
class VipRegistryNew(MemberRegistryNew):
    pass

vn1 = VipRegistryNew()
vn2 = VipRegistryNew()
print(vn1 is vn2)        # True — но по «случайной» причине:
                         # _instance унаследован от MemberRegistryNew и УЖЕ занят.
print(vn1 is n1)         # True ❌ — VipRegistryNew вернул ТОТ ЖЕ объект,
                         # что и MemberRegistryNew. Это скорее баг, чем фича.

"""
ВЫВОД ПРОГРАММЫ:

=== Метакласс ===
True
Реестр (метакласс): ['Иван', 'Мария']
Реестр (метакласс): ['Иван', 'Мария']
True
False

=== __new__ ===
True
Реестр (__new__): ['Пётр', 'Ольга']
Реестр (__new__): ['Пётр', 'Ольга']
True
True
"""

"""
СРАВНЕНИЕ ДВУХ ПОДХОДОВ:

┌──────────────────────────┬─────────────────────────┬─────────────────────────┐
│ Критерий                 │ Метакласс SingletonMeta │ __new__ в каждом классе │
├──────────────────────────┼─────────────────────────┼─────────────────────────┤
│ Где живёт логика         │ В ОДНОМ метаклассе      │ В КАЖДОМ классе —       │
│                          │                         │ копипаста               │
│ Наследники               │ Автоматически получают  │ Надо думать: _instance  │
│                          │ свой синглтон           │ унаследован и уже занят │
│ Читаемость               │ Сложнее: «магия»        │ Проще: видно __new__    │
│ Явность                  │ Неявно: класс помечен   │ Явно: класс сам хранит  │
│                          │ metaclass=...           │ _instance               │
│ __init__ вызывается      │ ТОЛЬКО один раз         │ Каждый раз — нужна      │
│                          │ (внутри __call__)       │ защита _initialized     │
│ Совместимость с другими  │ Конфликты, если у класса│ Никаких конфликтов      │
│ метаклассами             │ уже есть метакласс      │                         │
│ Много синглтонов         │ Одна строка на класс:   │ 5 строк × N классов     │
│                          │ metaclass=SingletonMeta │                         │
│ Тестируемость            │ Одинаково плохо         │ Одинаково плохо         │
└──────────────────────────┴─────────────────────────┴─────────────────────────┘

ПОЧЕМУ __init__ ВЕДЁТ СЕБЯ ПО-РАЗНОМУ:

Метакласс-подход перехватывает САМ ФАКТ создания через __call__:
    r1 = MemberRegistryMeta("аргумент")
    → SingletonMeta.__call__ → если объекта нет, вызывает
      type.__call__, который делает __new__ + __init__.
    → при повторном вызове __call__ просто возвращает готовый объект,
      НЕ вызывая __new__ и __init__ вообще.

__new__-подход перехватывает только __new__:
    n1 = MemberRegistryNew()
    → Python вызывает MemberRegistryNew.__new__ → возвращает существующий
      объект.
    → ЗАТЕМ Python ВСЕГДА вызывает __init__ у возвращённого объекта.
    → Значит, __init__ сработает ВТОРОЙ раз, если не защититься.

Именно поэтому в __new__-варианте приходится писать:

    if not hasattr(self, "_initialized"):
        self.members = []
        self._initialized = True

Иначе при каждом MemberRegistryNew() список members бы обнулялся.

ЭТО ВАЖНОЕ РАЗЛИЧИЕ: метакласс даёт «чистый» синглтон без хаков
в __init__, а __new__ требует аккуратной защиты.

ПРОБЛЕМА НАСЛЕДНИКОВ У __new__-ВАРИАНТА:

    class VipRegistryNew(MemberRegistryNew):
        pass

Атрибут _instance унаследован от MemberRegistryNew. Если MemberRegistryNew
уже создан, то _instance у него не None — и VipRegistryNew вернёт
ТОТ ЖЕ САМЫЙ объект MemberRegistryNew. Это, скорее всего, не то,
что ожидалось.

Чтобы починить, нужно в каждом наследнике СБРАСЫВАТЬ _instance:

    class VipRegistryNew(MemberRegistryNew):
        _instance = None      # ← иначе унаследуется занятый _instance

Или хранить _instances как словарь {cls: instance} — но тогда
мы фактически воспроизводим логику метакласса.

В метакласс-варианте этой проблемы нет: _instances — словарь по КЛЮЧУ-КЛАССУ,
поэтому у каждого класса свой синглтон автоматически.

КОГДА SINGLETON ПОЛЕЗЕН:

1. КОНФИГ / НАСТРОЙКИ ПРИЛОЖЕНИЯ.
       config = AppConfig()
   Один источник правды. Все модули читают одни и те же настройки.
   Альтернатива — модуль с константами, часто даже лучше.

2. ЛОГГЕР.
       log = Logger()
   Один логгер на приложение: общий уровень логирования,
   общий обработчик, единый формат вывода.

3. КЭШ.
       cache = Cache()
   Один кэш, общий для всех модулей: попадание из одного модуля
   видно другому.

4. ПУЛ РЕСУРСОВ / ПОДКЛЮЧЕНИЕ К БД.
   Один пул соединений, один клиент к внешнему сервису.

5. РЕЕСТР.
   Реестр клиентов, плагинов, обработчиков — то, что должно быть
   в единственном экземпляре по смыслу задачи.

6. ЖЕЛЕЗНЫЕ РЕСУРСЫ.
   Один физический принтер, одно устройство, один файл, к которому
   нельзя открыть два эксклюзивных доступа.

КОГДА SINGLETON ВРЕДЕН:

1. ТЕСТИРОВАНИЕ.
   Состояние сохраняется между тестами. Один тест изменил конфиг —
   другой тест упал. Приходится писать setUp-сбросы:

       def setUp(self):
           MemberRegistry._instance = None   # ← костыль

   Или использовать фикстуры, чтобы обнулять синглтон перед каждым тестом.
   И то и другое — «запах».

2. СКРЫТЫЕ ЗАВИСИМОСТИ.
   Функция берёт Logger() прямо внутри:

       def process():
           log = Logger()          # ← откуда взялся? неявная зависимость
           log.write(...)

   Внешне функция «чистая», а на деле зависит от глобального
   состояния. Тестировать и переиспользовать сложно.

   Лучше передавать явно (инъекция зависимости):

       def process(log):
           log.write(...)

3. НЕВОЗМОЖНОСТЬ ПОДМЕНИТЬ РЕАЛИЗАЦИЮ.
   Захотел заменить боевой логгер на «заглушку» в тестах — а подменить
   нечего: везде жёстко зашит один и тот же экземпляр.

   Решение — программировать против интерфейса и передавать
   зависимость снаружи:

       class Service:
           def __init__(self, logger):
               self.logger = logger

4. МНОГОПОТОЧНОСТЬ.
   Наивный синглтон не потокобезопасен: два потока могут одновременно
   пройти проверку `if cls._instance is None` и создать два объекта.
   Нужна блокировка:

       import threading

       class SingletonMeta(type):
           _instances = {}
           _lock = threading.Lock()

           def __call__(cls, *args, **kwargs):
               with cls._lock:
                   if cls not in cls._instances:
                       cls._instances[cls] = super().__call__(*args, **kwargs)
               return cls._instances[cls]

   Но и это не спасёт от «состояния гонки» на уровне бизнес-логики:
   два потока могут одновременно менять общий список.

5. ГЛОБАЛЬНОЕ СОСТОЯНИЕ.
   По сути, синглтон — это «легализованная глобальная переменная».
   Чем больше их в проекте, тем труднее рассуждать о поведении
   программы и отслеживать, кто и когда изменил состояние.

6. НАРУШЕНИЕ ПРИНЦИПА ЕДИНСТВЕННОЙ ОТВЕТСТВЕННОСТИ.
   Класс становится ответственным и за свою логику, и за контроль
   количества экземпляров. Это две разные задачи.

КОГДА SINGLETON ОПРАВДАН, А КОГДА ЛУЧШЕ ЗАМЕНИТЬ:

Оправдан, если:
- объект действительно один по своей природе (физический ресурс);
- состояние глобально по смыслу (конфиг, логгер);
- создание дорого и должно происходить один раз (пул соединений).

Лучше заменить:
- на модуль с глобальными переменными — для конфига и констант;
- на инъекцию зависимостей — для логгера, кэша, клиентов сервисов;
- на фабрику с явным управлением жизненным циклом — для пулов;
- на Borg (моносостояние) — когда нужны разные объекты, но общее
  состояние.

PYTHON-СПЕЦИФИЧНЫЕ АЛЬТЕРНАТИВЫ:

1. Модуль-синглтон — самый «питоничный» вариант:

       # config.py
       settings = {"debug": False}

   Импортируй откуда угодно — везде будет один и тот же объект,
   потому что модуль импортируется один раз.

2. Borg / Monostate — все экземпляры делят __dict__:

       class Borg:
           _shared = {}
           def __init__(self):
               self.__dict__ = Borg._shared

   Объектов много, состояние одно. Проще тестировать: можно
   создать новый «вид» и обнулить _shared.

3. @lru_cache как «синглтон для функций»:

       @lru_cache(maxsize=None)
       def get_config():
           return load_config()

   Первый вызов делает работу, последующие берут из кэша.

ИТОГ:

- Через метакласс: одна строка `metaclass=SingletonMeta` — и класс
  стал синглтоном; наследники автоматически получают свои синглтоны;
  __init__ вызывается один раз; но «магия» неявна и конфликтует
  с другими метаклассами.
- Через __new__: проще читать, но нужна защита __init__ от повторного
  вызова, и наследники легко ломаются из-за унаследованного _instance.
- Singleton полезен для конфига, логгера, кэша, пула ресурсов — когда
  объект действительно один по природе.
- Singleton вреден, когда скрывает зависимости, мешает тестировать,
  подменять реализацию и работать в многопоточном окружении.
- В Python часто достаточно модуля или @lru_cache — прежде чем
  писать синглтон, подумай об этих альтернативах.

ЗОЛОТОЕ ПРАВИЛО: «Синглтон — как соль: чуть-чуть улучшает вкус,
а много — портит блюдо». Применяй точечно, а не как универсальный
способ шарить данные.
"""

print()
```
