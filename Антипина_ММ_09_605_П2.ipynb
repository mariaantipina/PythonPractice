Задача 2: Создайте класс Document, который содержит название, автора и содержимое. Реализуйте методы update_content(new_content), который обновляет содержимое документа, и display_info(), который выводит полную информацию о документе.


```python
class Document:
    def __init__(self, title: str, author: str, content: str):
        self.title = title
        self.author = author
        self.content = content

    def update_content(self, new_content: str):
        self.content = new_content
        print(f"Содержимое документа «{self.title}» обновлено.")

    def display_info(self):
        """Выводит полную информацию о документе."""
        print(f"Название: {self.title}")
        print(f"Автор:    {self.author}")
        print(f"Содержимое: {self.content if self.content != "" else '(пусто)'}")
# Пример использования
if __name__ == "__main__":
    doc = Document("Название", "Мария", "Hello, world")
    doc.display_info()

    doc.update_content("Hello, world!!!")
    doc.display_info()
```

Задача 2. Определите класс Recipe, который наследуется от класса Document. Класс Recipe должен добавлять поля ингредиенты и время приготовления, а также методы для обновления времени приготовления и отображения информации о рецепте.


```python
class Recipe(Document):
    def __init__(self, title: str, author: str, content: str, ingredients: str, cooking_time: int):
        # Вызов конструктора родительского класса
        super().__init__(title, author, content)
        self.ingredients = ingredients
        self.cooking_time = cooking_time

    def update_cooking_time(self, new_time: int):
        self.cooking_time = new_time
        print(f"Время приготовления рецепта «{self.title}» обновлено: {new_time} мин.")

    def display_info(self):
        print(f"Рецепт:   {self.title}")
        print(f"Автор:    {self.author}")
        print(f"Время приготовления: {self.cooking_time} мин.")
        print(f"Ингредиенты:   {self.ingredients}")
        print(f"Инструкция: {self.content if self.content != "" else '(пусто)'}")

if __name__ == "__main__":
    recipe = Recipe(
        title="Борщ",
        author="Мария",
        content="1. Сварить бульон с мясом. 2. Добавить все остальные ингредиенты. 3. Варить 60 минут.",
        ingredients=["Свёкла", "Капуста", "Картофель", "Мясо", "Лук", "Морковь"],
        cooking_time=120
    )

    recipe.display_info()

    # Обновляем время приготовления
    recipe.update_cooking_time(160)
    recipe.display_info()

    # Можем также использовать методы родителя
    recipe.update_content("1. Сварить бульон. 2. Добавить овощи. 3. Варить 90 минут.")
    recipe.display_info()
```

Задача 2. Создать класс Recipe, у которого поля name, ingredients, cookingTime будут приватными. Реализовать методы для безопасного доступа и изменения этих полей, чтобы контролировать изменения ингредиентов и времени приготовления.


```python
class Recipe(Document):
    def __init__(self, title: str, author: str, content: str, ingredients: str, cooking_time: int):
        super().__init__(title, author, content)
        self.__ingredients = ingredients
        self.__cooking_time = cooking_time

    @property
    def ingredients(self):
        return self.__ingredients

    @ingredients.setter
    def ingredients(self, new_ingredients: str):
        self.__ingredients = new_ingredients.strip()

    @property
    def cooking_time(self):
        return self.__cooking_time

    @cooking_time.setter
    def cooking_time(self, new_time: int):
        self.__cooking_time = new_time

    def update_cooking_time(self, new_time: int):
        self.cooking_time = new_time
        print(f"Время приготовления рецепта «{self.title}» обновлено: {new_time} мин.")

    def display_info(self):
        print(f"Рецепт:   {self.title}")
        print(f"Автор:    {self.author}")
        print(f"Время приготовления: {self.cooking_time} мин.")
        print(f"Ингредиенты:   {self.ingredients}")
        print(f"Инструкция: {self.content if self.content != '' else '(пусто)'}")


if __name__ == "__main__":
    recipe = Recipe(
        title="Борщ",
        author="Мария",
        content="1. Сварить бульон с мясом. 2. Добавить все остальные ингредиенты. 3. Варить 60 минут.",
        ingredients="Свёкла, Капуста, Картофель, Мясо, Лук, Морковь",
        cooking_time=120
    )

    recipe.display_info()

    recipe.update_cooking_time(160)
    recipe.display_info()

    recipe.update_content("1. Сварить бульон. 2. Добавить овощи. 3. Варить 90 минут.")
    recipe.display_info()
```
