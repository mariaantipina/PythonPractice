```python
# Задача 3.
from math import sqrt
def perimetr_and_S_triangle(x1, y1, x2, y2, x3, y3):
 # длины сторон
 side_a = sqrt((x2 - x1)**2 + (y2 - y1)**2)
 side_b = sqrt((x3 - x2)**2 + (y3 - y2)**2)
 side_c = sqrt((x1 - x3)**2 + (y1 - y3)**2)

 perimetr = side_a + side_b + side_c
 s = 0.5 * abs(x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))
 return perimetr, s

# Ввод координат
x1, y1 = 0, 5
x2, y2 = 3, 1
x3, y3 = 2, 4

perimetr, s = perimetr_and_S_triangle(x1, y1, x2, y2, x3, y3)

print(f"Координаты вершин: ({x1}, {y1}), ({x2}, {y2}), ({x3}, {y3})")
print(f"Периметр треугольника: {perimetr}")
print(f"Площадь треугольника: {s}")
```
